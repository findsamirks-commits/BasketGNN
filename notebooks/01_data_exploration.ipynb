Analyzes quick-commerce user basket distributions, item co-purchase degrees, and connectivity sparsity.

```python
import os
import sys
import json
import pandas as pd
import numpy as np

# Add project root to sys.path
sys.path.insert(0, os.path.abspath('..'))
from src.data import load_dataset

data = load_dataset('../data/processed')
print(f"Users: {data['num_users']} | Items: {data['num_items']} | Interactions: {len(data['train_pairs'])}")
```

### 1. Catalog & Category Composition

```python
with open('../data/processed/catalog.json', 'r') as f:
    catalog = json.load(f)
df_cat = pd.DataFrame(catalog)
print(df_cat['category'].value_counts())
```

### 2. Interaction Degree Distribution

```python
df_inter = pd.read_csv('../data/processed/interactions.csv')
user_degrees = df_inter.groupby('user_id')['item_id'].count()
print(f"User interaction degree: mean={user_degrees.mean():.1f}, median={user_degrees.median():.1f}, max={user_degrees.max()}")

item_degrees = df_inter.groupby('item_id')['user_id'].count()
print(f"Item interaction degree: mean={item_degrees.mean():.1f}, median={item_degrees.median():.1f}, max={item_degrees.max()}")
```
